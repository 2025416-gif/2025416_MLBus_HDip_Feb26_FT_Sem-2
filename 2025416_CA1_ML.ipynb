## Introduction

This notebook covers data preparation, exploratory analysis, a comparison of two clustering algorithms, and the evaluation of ARIMA and SARIMA forecasting models.

Bicycle-sharing systems serve as an alternative mode of urban transportation. The number of rentals can fluctuate over time, influenced by factors such as weather conditions and the day of the week. In this project, I analyze these patterns using historical data from Capital Bikeshare, provided by the UCI (Fanaee-T, 2013).

I chose this topic because it allows for the application of machine learning techniques to an everyday scenario, with results that can be interpreted without the need for a complex business context. The objective is to investigate how days can be grouped based on their characteristics and to determine the extent to which the total daily number of rentals can be predicted.
